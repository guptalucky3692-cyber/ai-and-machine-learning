## ***LIMITS*** Where is function defined 
explain equation y = 3x******2/x******2 -4

```python
import sympy
from sympy import symbols
from sympy.solvers import solve
```

```python
x = symbols("x")
s = x**2 -4
t = 3*x**2
f = s/t
print('x =', solve(f,x))
```

***limit appproching left to right***

```python
x = 1
h = 2
x_right = (3*(x+h)**2/(x+h)**2 -4 )
x_left = (3*(x-h)**2/(x-h)**2 -4 )
print("x_right =", x_right)
print("x_left = ",x_left)
if round(x_left) != round(x_left):
    print("limit does not exist")
```
