```python
import pandas as pd

# 1. Convert DATE to datetime (or pass parse_dates=["DATE"] to pd.read_csv)
df_electric["DATE"] = pd.to_datetime(df_electric["DATE"])

# 2. Keep a real reference timestamp (don't overwrite it with the column)
today = pd.Timestamp.today()

# 3. Compute elapsed days from each historical date to today
days_since = (today - df_electric["DATE"]).dt.days
print(days_since.head())
```

```python
import pandas as pd

# Assuming `df_shampoo` is already loaded from sales-of-shampoo-over-a-three-ye.csv
# Inspect column names first if needed: print(df_shampoo.columns)
date_col = df_shampoo.columns[0]  # Often 'Month' in this dataset

# Convert to datetime safely
dt = pd.to_datetime(df_shampoo[date_col], errors="coerce")

# 1. Numeric month code (1 to 12)
df_shampoo["month"] = dt.dt.month

# 2. Month name (e.g., 'January', 'February') and short name ('Jan', 'Feb')
df_shampoo["month_name"] = dt.dt.month_name()
df_shampoo["month_abbr"] = dt.dt.strftime("%b")

# 3. Year-Month code (e.g., '1-01' / 'YYYY-MM') to distinguish repeated months across years
df_shampoo["year_month"] = dt.dt.to_period("M")

print(df_shampoo[[date_col, "month", "month_abbr", "year_month"]].head(10))
```
