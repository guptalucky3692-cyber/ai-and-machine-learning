```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df =  pd.read_csv('aug_train.csv')
df
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
flights = sns.load_dataset('flights')
iris = sns.load_dataset("iris")
```

```python
iris
```

> ## Scattter plot numerical numerical

```python
plt.figure(figsize=(6,5),facecolor='yellow')
sns.set_theme('poster')
sns.set_style('darkgrid')
sns.set_palette('rainbow')
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],color = "#CAF300")
```

> ## Bar plot ( Numerical - Categorical)

```python
df.head()
```

```python
plt.figure(figsize=(6,5),facecolor="#00DDFF")
sns.set_theme('poster')
sns.set_style('darkgrid')
sns.set_palette('rainbow')
sns.barplot(y= df['training_hours'].head(),x=df['experience'].head(),palette='rainbow',hue=df['target'])
```

> ## boxplot (numerical -   Categorical)

```python
plt.figure(figsize=(6,5),facecolor='orange')
sns.set_theme('notebook')
sns.set_style('darkgrid')
sns.set_palette('rainbow')
sns.boxplot(x=df['gender'],y=df['experience'],hue= df['gender'])
```

```python
plt.figure(figsize=(6,5),facecolor="#00D3F8")
sns.set_theme('poster')
sns.set_style('darkgrid')
sns.set_palette('rainbow')
sns.heatmap(pd.crosstab(df['education_level'],df['gender']),cmap='coolwarm')
sns.set_style('ticks')
```

> ## cluster map (categorical - Categorical)

```python
plt.figure(figsize=(6,5),facecolor='red')
sns.set_theme('notebook')
sns.set_style('darkgrid')
sns.set_palette('rainbow')
sns.clustermap(pd.crosstab(df['enrolled_university'],df['company_type']),cmap="coolwarm")
```
