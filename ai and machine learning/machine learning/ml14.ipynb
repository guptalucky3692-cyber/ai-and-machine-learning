## Mixing values

```python
import pandas as pd

url = "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day33-handling-mixed-variables/titanic.csv"
df = pd.read_csv(url)

print(df.head())
```

```python
df
```

```python
df['number'].unique()
```

```python
fig =  df['number'].value_counts().plot.bar()
fig.set_title('passanger travelling with')
```

```python
import numpy as np
df['number_numerical']= pd.to_numeric(df['number'],errors='coerce',downcast='integer')
df['number_categerical']= np.where(df['number_numerical'].isnull(),df['number'],np.nan)
```

```python
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['cabin_num']=df['Cabin'].str.extract('(\d+)')
df['Cabin_cat']= df['Cabin'].str[0]
df.head()
```

```python
df['Cabin_cat'].value_counts().plot(kind='bar')
```
